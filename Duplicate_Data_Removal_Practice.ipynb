**Dataset:** Employee dataset containing duplicate records

## Objective
Identify duplicate records, remove them, and compare row counts.

```python
import pandas as pd
df = pd.read_csv('employees_with_duplicates.csv')
display(df)
```

## 1. Count duplicate records

```python
duplicate_mask = df.duplicated()
print('Number of duplicate rows:', duplicate_mask.sum())
display(df[duplicate_mask])
```

## 2. Remove duplicate records

```python
cleaned_df = df.drop_duplicates()
print('Rows before cleaning:', len(df))
print('Rows after cleaning:', len(cleaned_df))
print('Rows removed:', len(df)-len(cleaned_df))
display(cleaned_df)
```

## 3. Check for remaining duplicates

```python
print('Remaining duplicate rows:', cleaned_df.duplicated().sum())
```

## 4. Remove duplicates based on employee ID

```python
cleaned_by_id = df.drop_duplicates(subset=['employee_id'])
print('Rows after keeping one row per employee ID:', len(cleaned_by_id))
display(cleaned_by_id)
```

## 5. Save cleaned dataset

```python
cleaned_df.to_csv('employees_cleaned.csv', index=False)
print('Saved employees_cleaned.csv')
```

## Key Observations
1. The original dataset contains 12 rows.
2. Some employee records are repeated exactly.
3. `duplicated()` identifies repeats after the first occurrence.
4. `drop_duplicates()` removes repeated rows and keeps the first occurrence by default.
5. Before-and-after row counts show how many records were removed.
6. Matching only employee IDs is a different rule from comparing every column.
